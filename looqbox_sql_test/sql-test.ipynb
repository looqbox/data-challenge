### SQL test

1) What are the 10 most expensive products in the company?

```python
query_sql_test_1 = "SELECT PRODUCT_NAME, PRODUCT_VAL FROM data_product ORDER BY PRODUCT_VAL DESC LIMIT 10;"

df_sql_test_1 = pd.read_sql(query_sql_test_1, engine)
print(df_sql_test_1)
```

2) What sections do the 'BEBIDAS' and 'PADARIA' departments have?

```python
query_sql_test_2 = "SELECT DISTINCT DEP_NAME, SECTION_NAME FROM data_product WHERE DEP_NAME IN ('BEBIDAS', 'PADARIA');"

df_sql_test_2 = pd.read_sql(query_sql_test_2, engine)
print(df_sql_test_2)
```

3) What was the total sale of products (in $) of each Business Area in the first quarter of 2019?

```python
query_sql_test_3 = "SELECT cad.BUSINESS_NAME, SUM(sales.SALES_VALUE) AS TOTAL_SALES FROM data_store_cad cad INNER JOIN data_store_sales sales ON cad.STORE_CODE = sales.STORE_CODE WHERE sales.DATE BETWEEN '2019-01-01' AND '2019-03-31' GROUP BY cad.BUSINESS_NAME;"

df_sql_test_3 = pd.read_sql(query_sql_test_3, engine)
print(df_sql_test_3)
```
