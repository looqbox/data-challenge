## SQL Question 3

```python
query = """
SELECT
    sc.BUSINESS_NAME,
    SUM(ps.SALES_VALUE) AS TOTAL_SALES
FROM data_product_sales ps
INNER JOIN data_store_cad sc
    ON ps.STORE_CODE = sc.STORE_CODE
WHERE ps.DATE BETWEEN '2019-01-01' AND '2019-03-31'
GROUP BY sc.BUSINESS_NAME
ORDER BY TOTAL_SALES DESC;
"""

q3 = pd.read_sql(query, conn)

q3["TOTAL_SALES"] = q3["TOTAL_SALES"].map(lambda x: f"R$ {x:,.2f}")

print(q3)
```

```python
conn = pymysql.connect(
    host="35.199.115.174",
    user="looqbox-challenge",
    password="looq-challenge",
    database="looqbox-challenge",
    port=3306
)

query1 = """
SELECT
      STORE_CODE,
      STORE_NAME,
      START_DATE,
      END_DATE,
      BUSINESS_NAME,
      BUSINESS_CODE
FROM data_store_cad
"""

query2 = """
SELECT
        STORE_CODE,
        DATE,
        SALES_VALUE,
        SALES_QTY
FROM data_store_sales
WHERE DATE BETWEEN '2019-01-01' AND '2019-12-31'
"""

df_store = pd.read_sql(query1, conn)
df_sales = pd.read_sql(query2, conn)
```

```python
df_sales["DATE"] = pd.to_datetime(df_sales["DATE"])

df_sales = df_sales[
    (df_sales["DATE"] >= "2019-10-01") &
    (df_sales["DATE"] <= "2019-12-31")
]
```

```python
df_grouped = df_sales.groupby("DATE")[["SALES_VALUE", "SALES_QTY"]].sum()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(df_grouped.index, df_grouped["SALES_VALUE"], label="Sales Value")
plt.plot(df_grouped.index, df_grouped["SALES_QTY"], label="Sales Quantity")

plt.title("Sales Performance (Q4 2019)")
plt.xlabel("Date")
plt.ylabel("Value")
plt.legend()
plt.xticks(rotation=45)

plt.show()
```

# CASE 3

```python
query3 = """
SELECT *
FROM IMDB_movies
"""

df_movies = pd.read_sql(query3, conn)
```

```python
df_movies.head()
df_movies.info()
df_movies.describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(df_movies["Rating"].dropna(), bins=20, edgecolor="black")

plt.title("Distribution of IMDB Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.show()
```

```python
df_genre = (
    df_movies.groupby("Genre")["Rating"]
    .mean()
    .sort_values()
)

plt.figure(figsize=(10,6))

df_genre.tail(15).plot(kind="barh")  # top 15 só

plt.title("Top Genres by Average Rating")
plt.xlabel("Average Rating")
plt.ylabel("Genre")

plt.grid(axis="x", alpha=0.3)

plt.show()
```
