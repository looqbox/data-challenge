## Caso 1 - Função dinâmica para consultar a tabela data_product_sales

```python
# link do notebook com o sistema
import sys
import os

# raiz do projeto
if root not in sys.path:
    sys.path.insert(0, root)

print(sys.path[0])  # confirma
```

```python
# importações

from src.models.schemas.product_sales_schema import ProductSalesSchema
from src.models.repositories.sales_repository import SalesRepository
from typing import List
import pandas as pd

repo = SalesRepository()
schema = ProductSalesSchema()
```

```python
# função retrieve_data 

def retrieve_data(product_code:int = None, store_code:int = None, data_range:List = None) -> pd.DataFrame:
    """
        Recupera dados da tabela data_product_sales com filtros opcionais.

        Parâmetros

        product_code: int -> código do produto, se NONE retorna todos

        store_code: int -> código da loja, se NONE rotorna todos

        data_range: list -> intervalo de datas, aceita NONE para intervalos abertos
    
    """

    product_code = schema.product_validator(product_code)
    store_code = schema.store_validator(store_code)
    data_range = schema.data_validator(data_range)

    response = repo.read_db(product_code=product_code, store_code= store_code, data_range= data_range)

    return response 
```

```python
# uso da função - sem filtros

base = retrieve_data()
base.describe(), base.info
```

```python
# uso da função - só o produto

base = retrieve_data(product_code= 18)
print(base.describe())
print(base.info)
```

```python
# uso da função - só a loja 
base = retrieve_data(store_code=1)
print(base.describe())
print(base.info)
```

```python
# uso da função - datas - intervalo fechado
base = retrieve_data(data_range=["2019-01-01","2019-01-31"])
print(base.describe())
print(base.info)
```

```python
# uso da função - datas - intervalo aberto
base = retrieve_data(data_range=[ None,"2019-01-31"])
print(base.describe())
print(base.info)
```

```python
# uso da função - todos os filtros
base = retrieve_data(product_code= 18, store_code=1, data_range=["2019-01-01","2019-01-31"])
print(base.describe())
print(base.info)
```

## Decisões técnicas

- Parâmetros opcionais com default `None` — qualquer combinação de filtros é válida
- Validação isolada no Schema antes de chegar ao Repository
- Datas entram como string ISO e são convertidas para `datetime.date` internamente
- Filtro de data suporta intervalo aberto (só início ou só fim)
- Queries montadas com SQLAlchemy Core — sem concatenação de string, sem risco de SQL Injection
- IA utilizada para: revisão de bugs nos validators e estrutura dos testes pytest
